```python
numbers = [1,2,3,4,5,6,7]
remaining_numbers = [8,9,10]

numbers.append(remaining_numbers)

print(numbers)
```

```python
numbers = [1,2,3,4,5,6,7]
remaining_numbers = [8,9,10]

numbers.extend(remaining_numbers)

print(numbers)
```

```python
# at specific index

numbers.insert(2,89)

print(numbers)
```

```python
numbers.remove(89)

print(numbers)
```

```python
numbers.append(7)
print(numbers)

numbers.remove(7)
print(numbers)
```

```python
numbers.pop(1)

print(numbers)
```

```python
numbers.pop()

print(numbers)
```

```python
numbers.clear()

print(numbers)
```

```python
numbers = [1,2,3,7,8,3,2,4,5,6,7,0,8,9,10]

numbers.sort()

print(numbers)
```

```python
# sorted() function which works for any iterable and returns a new sorted list instead of modifying the original list.
numbers=[2,4,2,4,32,4,3,2,4,3,4,4,3,5,4,56567,6785,7,5,87]
newww = sorted(numbers)

print(newww)
```

```python
numbers=[2,4,2,4,32,4,3,2,4,3,4,4,3,5,4,56567,6785,7,5,87]

numbers.reverse()

print(numbers)
```
